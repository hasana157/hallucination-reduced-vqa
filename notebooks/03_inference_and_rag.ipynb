# Module 3: Inference & Grounding Pipeline (RAG + VCD)

This notebook demonstrates the end-to-end inference pipeline supporting Mode A (Baseline), Mode B (QLoRA Only), and Mode C (QLoRA + RAG + VCD).

```python
# Setup paths
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

from src.inference import InferencePipeline, InferenceConfig
from PIL import Image
import torch

print(f"CUDA Available: {torch.cuda.is_available()}")
```

```python
# Initialize pipeline config
config = InferenceConfig(
    base_model="Qwen/Qwen2-VL-2B-Instruct",
    lora_path="checkpoints/lora_weights",
    faiss_index_path="data/faiss_index",
    entropy_threshold=0.55,
    top_k=5,
    vcd_alpha=0.5,
    blur_radius=15
)

pipeline = InferencePipeline(config)
```

```python
# Test inference across configurations A, B, C
sample_image = Image.new("RGB", (224, 224), color=(100, 150, 200))
question = "What is present in this image?"

for mode in ["A", "B", "C"]:
    result = pipeline.run(sample_image, question, mode=mode)
    print(f"Mode {mode}: Answer='{result.answer}', Latency={result.latency_seconds:.2f}s, Memory={result.gpu_memory_gb:.2f}GB, RAG={result.rag_triggered}")
```
